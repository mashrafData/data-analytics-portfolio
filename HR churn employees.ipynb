```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go 
import plotly.express as px 
df = pd.read_csv('HR-Employee-Attrition.csv')
```

```python
#Attrition Distribution
fig = px.pie(df, names='Attrition', title='Attrition Distribution')
fig.show()
```

```python
#Attrition by Department
```

```python

dep = df.groupby(['Department','Attrition']).size().reset_index(name='Count')

fig = px.bar(dep, x='Department', y='Count',
             color='Attrition',
             barmode='group',
             title='Attrition by Department')

fig.show()
```

```python
#Education Field vs Attrition
edu_field = df.groupby(['EducationField','Attrition']).size().reset_index(name='Count')

fig = px.bar(edu_field, x='EducationField', y='Count',
             color='Attrition',
             title='Attrition by Education Field')

fig.show()
```

```python
# Attrition by Marital Status

marital = df.groupby(['MaritalStatus','Attrition']).size().reset_index(name='Count')

fig = px.bar(marital, x='MaritalStatus', y='Count',
             color='Attrition',
             title='Attrition by Marital Status')

fig.show()
```

```python
#Job Role vs Attrition

job = df.groupby(['JobRole','Attrition']).size().reset_index(name='Count')

fig = px.bar(job, x='JobRole', y='Count',
             color='Attrition',
             title='Attrition by Job Role')

fig.show()
```

```python
# Years at Company vs Attrition

fig = px.box(df, x='Attrition', y='YearsAtCompany',
             title='Years at Company vs Attrition')

fig.show()
```

```python
#Age vs Attrition

fig = px.histogram(df, x='Age', color='Attrition',
                   barmode='overlay',
                   title='Age Distribution vs Attrition')

fig.show()
```

```python
#Attrition by OverTime
ot = df.groupby(['OverTime','Attrition']).size().reset_index(name='Count')

fig = px.bar(ot, x='OverTime', y='Count',
             color='Attrition',
             title='Attrition by OverTime')

fig.show()
```

```python
#Salary vs Attrition


fig = px.box(df, x='Attrition', y='MonthlyIncome',
             title='Monthly Income vs Attrition')

fig.show()
```
